```python
orders = pd.DataFrame({
    "order_id": range(101, 113),

    "city": [
        "Lagos", "Abuja", "Lagos", "Kano",
        "Abuja", "Lagos", "Kano", "Abuja",
        "Lagos", "Kano", "Abuja", "Lagos"
    ],

    "channel": [
        "Online", "Store", "Online", "Store",
        "Online", "Store", "Online", "Online",
        "Store", "Store", "Online", "Online"
    ],

    "order_value": [
        2000, 4000, 3000, 3500,
        5000, 4500, 2500, 3000,
        80000, 4000, 5500, 3500
    ],

    "delivery_days": [
        2, 4, 3, 5, 2, 3,
        6, 2, 4, 5, 3, 2
    ],

    "age": [
        25, 34, 29, 41, None, 31,
        28, 35, 52, 46, 33, 27
    ]
})

assert orders["order_id"].is_unique

print(orders.shape)
print(orders.head())
print(orders.dtypes)
```

```python
data_path = Path("../data/processed/orders.csv")

orders.to_csv(data_path, index=False)

print("Dataset saved successfully!")
```

```python
print("Dataset Shape:")
print(orders.shape)

print("\nFirst 5 Rows:")
display(orders.head())

print("\nLast 5 Rows:")
display(orders.tail())

print("\nColumn Names:")
print(orders.columns)

print("\nData Types:")
print(orders.dtypes)
```

```python
print("Missing Values:")
print(orders.isna().sum())
```

```python
missing_percentage = (
    orders.isna().sum() / len(orders)
) * 100

print(missing_percentage)
```

```python
print("Total Duplicate Rows:")
print(orders.duplicated().sum())

print("\nDuplicate Order IDs:")
print(orders.duplicated(subset=["order_id"]).sum())
```

```python
print("Unique Values:")
print(orders.nunique(dropna=False))
```

```python
print(orders["city"].value_counts(dropna=False))
```

```python
print("Number of Orders:")
print(len(orders))

print("\nNumber of Available Ages:")
print(orders["age"].count())

print("\nMissing Ages:")
print(orders["age"].isna().sum())

print("\nMost Common City:")
print(orders["city"].mode())

print("\nCity Frequencies:")
print(orders["city"].value_counts())
```

```python
channel_percentage = (
    orders["channel"]
    .value_counts(normalize=True)
    .mul(100)
)

print(channel_percentage)
```

```python
x = orders["order_value"]

print("Mean:", x.mean())

print("Median:", x.median())

print("Mode:")
print(x.mode())

print("Standard Deviation:", x.std())

print("\nQuartiles:")
print(x.quantile([0.25, 0.50, 0.75, 0.95]))

print("\nComplete Statistical Summary:")
print(x.describe())
```

```python
print(orders[[
    "order_value",
    "delivery_days",
    "age"
]].describe())
```

```python
summary = orders.groupby(
    "city",
    as_index=False
).agg(
    orders=("order_id", "nunique"),
    total_revenue=("order_value", "sum"),
    mean_order=("order_value", "mean"),
    median_order=("order_value", "median")
)

display(summary.sort_values(
    "total_revenue",
    ascending=False
))
```

```python
channel_summary = orders.groupby(
    "channel"
)["order_value"].describe()

display(channel_summary)
```

```python
age_summary = orders.groupby("channel").agg(
    average_age=("age", "mean"),
    available_ages=("age", "count")
)

display(age_summary)
```

```python
fig, ax = plt.subplots(figsize=(8, 4))

sns.histplot(
    data=orders,
    x="order_value",
    bins=8,
    ax=ax
)

ax.set(
    title="Order Value Distribution",
    xlabel="Order Value (NGN)",
    ylabel="Number of Orders"
)

plt.tight_layout()

plt.savefig(
    "../figures/histogram.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 4))

sns.histplot(
    data=orders,
    x="order_value",
    bins=[0, 2000, 4000, 6000, 10000, 100000],
    ax=ax
)

ax.set(
    title="Order Values Using Business-Defined Bins",
    xlabel="Order Value (NGN)",
    ylabel="Orders"
)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 3))

sns.boxplot(
    data=orders,
    x="order_value",
    ax=ax
)

ax.set(
    title="Order Value Spread and Potential Outliers",
    xlabel="Order Value (NGN)"
)

plt.tight_layout()

plt.savefig(
    "../figures/boxplot.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

```python
q1, q3 = orders["order_value"].quantile(
    [0.25, 0.75]
)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

flagged = orders.loc[
    (orders["order_value"] < lower) |
    (orders["order_value"] > upper)
]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)

print("Lower Boundary:", lower)
print("Upper Boundary:", upper)

print("\nPotential Outliers:")

display(flagged[[
    "order_id",
    "order_value"
]])
```

```python
print(
    "Mean with all observations:",
    orders["order_value"].mean()
)

print(
    "Mean without flagged observations:",
    orders.loc[
        ~orders.index.isin(flagged.index),
        "order_value"
    ].mean()
)

print(
    "Median with all observations:",
    orders["order_value"].median()
)
```

```python
counts = (
    orders["city"]
    .value_counts()
    .rename_axis("city")
    .reset_index(name="orders")
)

fig, ax = plt.subplots(figsize=(7, 4))

sns.barplot(
    data=counts,
    x="city",
    y="orders",
    ax=ax
)

ax.set(
    title="Order Count by City",
    xlabel="City",
    ylabel="Number of Orders"
)

ax.set_ylim(bottom=0)

plt.tight_layout()

plt.savefig(
    "../figures/city_orders.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

```python
revenue = orders.groupby(
    "city",
    as_index=False
)["order_value"].sum()

fig, ax = plt.subplots(figsize=(7, 4))

sns.barplot(
    data=revenue,
    x="city",
    y="order_value",
    ax=ax
)

ax.set(
    title="Total Revenue by City",
    xlabel="City",
    ylabel="Revenue (NGN)"
)

ax.set_ylim(bottom=0)

plt.tight_layout()

plt.savefig(
    "../figures/city_revenue.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

```python
median_revenue = orders.groupby(
    "city",
    as_index=False
)["order_value"].median()

fig, ax = plt.subplots(figsize=(7, 4))

sns.barplot(
    data=median_revenue,
    x="city",
    y="order_value",
    ax=ax
)

ax.set(
    title="Median Order Value by City",
    xlabel="City",
    ylabel="Median Order Value (NGN)"
)

ax.set_ylim(bottom=0)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

sns.scatterplot(
    data=orders,
    x="delivery_days",
    y="order_value",
    hue="channel",
    s=85,
    ax=ax
)

ax.set(
    title="Delivery Time Versus Order Value",
    xlabel="Delivery Time (Days)",
    ylabel="Order Value (NGN)"
)

plt.tight_layout()

plt.savefig(
    "../figures/scatterplot.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

```python
print("Pearson Correlation:")

print(
    orders[[
        "delivery_days",
        "order_value"
    ]].corr(method="pearson")
)

print("\nSpearman Correlation:")

print(
    orders[[
        "delivery_days",
        "order_value"
    ]].corr(method="spearman")
)
```

```python
numeric = orders[[
    "order_value",
    "delivery_days",
    "age"
]]

corr = numeric.corr(numeric_only=True)

fig, ax = plt.subplots(figsize=(6, 4))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    ax=ax
)

ax.set(
    title="Correlation Among Numerical Variables"
)

plt.tight_layout()

plt.savefig(
    "../figures/correlation_heatmap.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

```python
cross = pd.crosstab(
    orders["city"],
    orders["channel"]
)

fig, ax = plt.subplots(figsize=(6, 4))

sns.heatmap(
    cross,
    annot=True,
    fmt="d",
    cmap="Blues",
    ax=ax
)

ax.set(
    title="Order Counts: City by Channel",
    xlabel="Sales Channel",
    ylabel="City"
)

plt.tight_layout()
plt.show()
```

```python
pct = pd.crosstab(
    orders["city"],
    orders["channel"],
    normalize="index"
).mul(100)

display(pct.round(1))
```
